```python
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input('Digite o segundo número: '))
operação = input('Digite a operação que deseja realizar: ')
if operação == 'soma':
    print('A soma entre os dois números é igual a:', numero1 + numero2)
elif operação == 'sutração':
    print('A subtração entre os dois números é igual a:', numero1 - numero2)
elif operação == 'multiplicação':
      print('A multiplicação entre os dois números é igual a:', numero1 * numero2)
elif operação == 'divisão':
        print('A divisão entre os dois números é igual a:', numero1 / numero2)
```
