
# Conheça as estruturas condicionais

```python
# if
if 12 > 10:
  print("Verdade")
else:
  print("Falso")
```

```python
#if/else
num1 = 12
num2 = 14
if num1 > num2:
  print("O num1 é o maior número")
elif num2 > num1:
  print("O num2 é o maior número")
else:
  print("Os dois são iguais")
```

```python
num1 = 10
num2 = 5
num3= 2
if (num1 > num2) and (num1 > num3):
  print("O num1 é o maior número")
else:
  print("o num1 não é o maior número")
```

#Laço de Repetição

```python
# For Range
for contador in range(10):
    print(contador)
```

```python
# For Range + Break
for contador in range(1000):
    print(contador)
    if contador == 5:
        break
```

```python
lista  = ['Teste1','Teste2','Teste3','Teste4']
```

```python
# For Simples
for val in lista:
      print(val)
```

```python
  # For com Enumerate
for i, valor in enumerate(lista):
      print(i, valor)
```

## For com Operadores Compostos

```python
lista2 = [2, 3, 4]
```

```python
for val in lista2:
    val += 1
    print(val)
```

```python
for val in lista2:
    val *= 2
    print(val)
```

## While

```python
val = 1
while (val < 6):
    val *=2
    print(val)
```

```python
# While com else (Multiplica-se 2,4,8....16 = não é valido por é maior que 10)
val = 1
while val < 5:
    val *=2
else:
    print('Valor de val é:', val)
```

#Exemplo de código executável (em Python) que ilustra boas práticas:


```python
nome_usuario = input("Por favor, insira seu nome: ")
print(f"Olá, {nome_usuario}! Bem-vindo ao nosso sistema.")

# Solicita a idade do usuário e converte para inteiro
idade_usuario = int(input("Por favor, insira sua idade: "))

# Verifica se o usuário é maior de idade
if idade_usuario >= 18:
    print("Você é maior de idade.")
else:
    print("Você é menor de idade.")
```
